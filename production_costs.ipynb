# Total production costs (KPI calculations)

Production costs per item of each use case, with State of the Art (SoA) technologies and with the Lasers4MaaS solution (L4M).
The data comes from `Input_data/KPI_calculations.xlsx`, sheets `Comp-usecase1` and `Comp-usecase2` (rows "Labour costs" to "Depreciation costs").
This notebook does not need Brightway.

```python
import plotly.graph_objects as go
from IPython.display import display
from l4m_lca import plots
from l4m_lca.costs import read_production_costs
```

```python
costs = read_production_costs()
costs.pivot_table(index=["use_case", "technology"], columns="cost", values="value", sort=False, margins=True,
                  margins_name="Total", aggfunc="sum").drop(index="Total").round(2)
```

A click on a legend item hides or shows that cost type. The totals and the y axes follow.

```python
fig = plots.production_cost_bars(costs)
display(fig)
go.Figure(fig).write_html("results/production_costs.html", include_plotlyjs="cdn")
# Image for a report (needs the kaleido package):
# plots.for_print(fig).write_image("results/production_costs.png", scale=2)
```
